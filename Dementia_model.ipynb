```
import tensorflow as tf
from tensorflow.keras import layers, models, regularizers, callbacks
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import matplotlib.pyplot as plt
import os
import zipfile
import sys
import numpy as np
import cv2
from sklearn.model_selection import train_test_split
from sklearn.utils import shuffle

# --- 1. Configuration ---

ZIP_FILE_NAME = 'Alzheimer_MRI_4_classes_dataset.zip'
DATA_DIR = 'Alzheimer_MRI_4_classes_dataset'
IMG_HEIGHT = 176
IMG_WIDTH = 208
BATCH_SIZE = 16
NUM_CLASSES = 4
EPOCHS = 30

# --- 2. Data Helper Functions ---

def load_data_split_and_balance(data_dir, img_height, img_width):
    """
    1. Loads ORIGINAL data.
    2. Splits into Train (80%) and Val (20%) - Stratified.
    3. BALANCES only the Training set (Oversampling minority classes).
    4. Leaves Validation set pure (unbalanced or original distribution).
    """
    print("--- Starting Data Pipeline ---")

    if not os.path.exists(data_dir):
        print(f"Error: {data_dir} not found.")
        sys.exit()

    class_names = sorted([d for d in os.listdir(data_dir) if os.path.isdir(os.path.join(data_dir, d))])
    print(f"Classes: {class_names}")

    # 1. Load All Original Images
    X_all = []
    y_all = []

    for label_idx, class_name in enumerate(class_names):
        class_path = os.path.join(data_dir, class_name)
        print(f"Loading {class_name}...")
        files = os.listdir(class_path)
        for fname in files:
            img_path = os.path.join(class_path, fname)
            try:
                img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
                if img is not None:
                    img = cv2.resize(img, (img_width, img_height))
                    img = img.astype('float32') / 255.0
                    X_all.append(img)
                    y_all.append(label_idx)
            except Exception:
                pass

    X_all = np.array(X_all).reshape(-1, img_height, img_width, 1)
    y_all = np.array(y_all)

    # 2. Split (Stratified to keep class ratios equal in both sets initially)
    print(f"Splitting data 80/20...")
    X_train, X_val, y_train, y_val = train_test_split(
        X_all, y_all, test_size=0.2, random_state=42, stratify=y_all
    )

    # 3. Balance Training Set ONLY
    print("Balancing Training Set...")
    X_train_balanced = []
    y_train_balanced = []

    # Find max class count in TRAINING set
    unique, counts = np.unique(y_train, return_counts=True)
    max_count = max(counts)
    print(f"  Target count per class: {max_count}")

    # Generator for MRI (Constant black background is CRITICAL)
    datagen = ImageDataGenerator(
        rotation_range=8,      # Slight rotation
        width_shift_range=0.05,
        height_shift_range=0.05,
        zoom_range=0.05,
        horizontal_flip=True,
        fill_mode='constant',  # Black background
        cval=0.0
    )

    for label_idx in range(NUM_CLASSES):
        # Get all images of this class from Training set
        indices = np.where(y_train == label_idx)[0]
        class_imgs = X_train[indices]

        # Add originals
        X_train_balanced.extend(class_imgs)
        y_train_balanced.extend([label_idx] * len(class_imgs))

        # Generate needed
        needed = max_count - len(class_imgs)
        if needed > 0:
            # Reshape for generator
            source = class_imgs
            gen = datagen.flow(source, batch_size=1, seed=42)
            for _ in range(needed):
                X_train_balanced.append(next(gen)[0])
                y_train_balanced.append(label_idx)

    X_train_balanced = np.array(X_train_balanced)
    y_train_balanced = np.array(y_train_balanced)

    # Shuffle Training Data
    X_train_balanced, y_train_balanced = shuffle(X_train_balanced, y_train_balanced, random_state=123)

    print(f"Final Training Shape: {X_train_balanced.shape} (Balanced)")
    print(f"Final Validation Shape: {X_val.shape} (Original Distribution)")

    return X_train_balanced, y_train_balanced, X_val, y_val

# --- 3. Run Data Prep ---

if not os.path.exists(DATA_DIR):
    if os.path.exists(ZIP_FILE_NAME):
        with zipfile.ZipFile(ZIP_FILE_NAME, 'r') as zip_ref:
            zip_ref.extractall('.')
    else:
        print("Zip file not found!")
        sys.exit()

X_train, y_train, X_val, y_val = load_data_split_and_balance(DATA_DIR, IMG_HEIGHT, IMG_WIDTH)

# Create Datasets
train_dataset = tf.data.Dataset.from_tensor_slices((X_train, y_train))
train_dataset = train_dataset.shuffle(2000).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

val_dataset = tf.data.Dataset.from_tensor_slices((X_val, y_val))
val_dataset = val_dataset.batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

# --- 4. Advanced Model Modules ---

def cbam_block(input_tensor, reduction_ratio=16, reg_val=0.001):
    """ CBAM with proper Keras/TF Integration """
    channels = input_tensor.shape[-1]

    # Channel Attention
    avg_pool = layers.GlobalAveragePooling2D()(input_tensor)
    max_pool = layers.GlobalMaxPooling2D()(input_tensor)

    shared_mlp = models.Sequential([
        layers.Dense(channels // reduction_ratio, activation='relu', kernel_regularizer=regularizers.l2(reg_val), use_bias=False),
        layers.Dense(channels, activation='sigmoid', kernel_regularizer=regularizers.l2(reg_val), use_bias=False)
    ])

    avg_pool = shared_mlp(avg_pool)
    max_pool = shared_mlp(max_pool)

    channel_attention = layers.Add()([avg_pool, max_pool])
    x = layers.Multiply()([input_tensor, channel_attention])

    # Spatial Attention (Wrapped in Lambda for Keras 3 compatibility)
    avg_pool = layers.Lambda(lambda t: tf.reduce_mean(t, axis=-1, keepdims=True))(x)
    max_pool = layers.Lambda(lambda t: tf.reduce_max(t, axis=-1, keepdims=True))(x)
    concat = layers.Concatenate(axis=-1)([avg_pool, max_pool])

    spatial_attention = layers.Conv2D(1, (7, 7), padding='same', activation='sigmoid',
                                      kernel_regularizer=regularizers.l2(reg_val))(concat)
    output = layers.Multiply()([x, spatial_attention])
    return output

def vision_transformer_block(inputs, num_heads, key_dim, ff_dim, dropout=0.1, reg_val=0.001):
    """ ViT Block with integer shaping """
    shape = inputs.shape
    h, w, c = shape[1], shape[2], shape[3]

    # Flatten with integer dimensions
    x = layers.Reshape((int(h) * int(w), int(c)))(inputs)

    # Attention
    x_norm = layers.LayerNormalization(epsilon=1e-6)(x)
    attn = layers.MultiHeadAttention(num_heads=num_heads, key_dim=key_dim, dropout=dropout)(x_norm, x_norm)
    x = layers.Add()([x, attn])

    # MLP
    x_norm = layers.LayerNormalization(epsilon=1e-6)(x)
    ffn = models.Sequential([
        layers.Dense(ff_dim, activation='relu', kernel_regularizer=regularizers.l2(reg_val)),
        layers.Dropout(dropout),
        layers.Dense(int(c), kernel_regularizer=regularizers.l2(reg_val))
    ])
    x_ffn = ffn(x_norm)
    x = layers.Add()([x, x_ffn])

    # Reshape back
    x = layers.Reshape((int(h), int(w), int(c)))(x)
    return x

# --- 5. Build Robust Hybrid Model ---

def build_hybrid_model():
    inputs = layers.Input(shape=(IMG_HEIGHT, IMG_WIDTH, 1))

    # Stronger initial regularization (noise)
    x = layers.GaussianNoise(0.01)(inputs)

    # Filters: 16 -> 32 -> 64 -> 128 -> 256
    filters_list = [16, 32, 64, 128, 256]
    # Increased regularization to fight overfitting
    L2_REG = 0.005

    for i, f in enumerate(filters_list):
        print(f"Building Block {i+1} ({f} filters)...")

        # Conv + BN + Activation
        x = layers.Conv2D(f, (3, 3), padding='same', use_bias=False, kernel_regularizer=regularizers.l2(L2_REG))(x)
        x = layers.BatchNormalization()(x)
        x = layers.Activation('relu')(x)

        # CBAM
        x = cbam_block(x, reg_val=L2_REG)

        # Pool
        x = layers.MaxPooling2D((2, 2))(x)

        # Conditional ViT (Blocks 3, 4, 5 only)
        if i >= 2:
            # Reduce ViT complexity to prevent memory overload and overfitting
            ff_dim = f * 2
            x = vision_transformer_block(x, num_heads=2, key_dim=f, ff_dim=ff_dim, dropout=0.3, reg_val=L2_REG)

        # Dropout
        x = layers.Dropout(0.3)(x)

    # Head
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dense(256, activation='relu', kernel_regularizer=regularizers.l2(L2_REG))(x)
    x = layers.Dropout(0.5)(x)
    outputs = layers.Dense(NUM_CLASSES, activation='softmax')(x)

    model = models.Model(inputs, outputs, name="High_Accuracy_Hybrid_Net")
    # Using a slightly lower learning rate for stability
    model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])
    return model

# --- 6. Train ---

tf.keras.backend.clear_session()
model = build_hybrid_model()
model.summary()

callbacks_list = [
    callbacks.ReduceLROnPlateau(
        monitor='val_loss', factor=0.5, patience=3, min_lr=1e-7, verbose=1
    ),
    callbacks.EarlyStopping(
        monitor='val_accuracy', patience=8, restore_best_weights=True, verbose=1
    )
]

print("\nStarting Optimized Training...")
history = model.fit(
    train_dataset,
    validation_data=val_dataset,
    epochs=EPOCHS,
    callbacks=callbacks_list,
    verbose=1
)

# --- 7. Plot Results ---

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Train Acc')
plt.plot(history.history['val_accuracy'], label='Val Acc')
plt.title('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Val Loss')
plt.title('Loss')
plt.legend()

plt.savefig('high_accuracy_results.png')
plt.show()

# Final Test
val_loss, val_acc = model.evaluate(val_dataset)
print(f"\nFINAL VALIDATION ACCURACY: {val_acc*100:.2f}%")
```
