# Data Augmentation

This notebook demonstrates the data augmentation pipeline used for the Alzheimer's MRI dataset.

The augmentations are applied only during training to improve model generalization and reduce overfitting.

The following augmentations are used:

- Rotation
- Brightness & Contrast Adjustment
- Gaussian Noise
- Affine Transformation (Translation & Scaling)

```python
df = load_dataset(DATASET_PATH)
df.head()
```

```python
augmented = augment_image(image)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(15, 10))

for i in range(6):

    augmented = augment_image(image)

    plt.subplot(2, 3, i + 1)
    plt.imshow(augmented)
    plt.axis("off")
    plt.title(f"Sample {i+1}")

plt.tight_layout()
plt.show()
```
