# Model Training – EfficientNetV2 Baseline

## Objective

Train the EfficientNetV2B0 baseline model using the production training pipeline.

The model is trained using:

- TensorFlow
- Transfer Learning
- EfficientNetV2B0
- Adam Optimizer
- Sparse Categorical Crossentropy
- EarlyStopping
- ModelCheckpoint
- ReduceLROnPlateau
- CSVLogger

This notebook summarizes the training process and results.

```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().parent

history_path = PROJECT_ROOT / "outputs" / "history.csv"

history = pd.read_csv(history_path)

history.head()
```

```python
log_path = PROJECT_ROOT / "outputs" / "training_log.csv"

training_log = pd.read_csv(log_path)

training_log.head()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,5))

plt.plot(history["accuracy"], label="Train")

plt.plot(history["val_accuracy"], label="Validation")

plt.title("Training vs Validation Accuracy")

plt.xlabel("Epoch")

plt.ylabel("Accuracy")

plt.legend()

plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.plot(history["loss"], label="Train")

plt.plot(history["val_loss"], label="Validation")

plt.title("Training vs Validation Loss")

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.legend()

plt.grid(True)

plt.show()
```

```python
best_epoch = history["val_accuracy"].idxmax()

history.iloc[best_epoch]
```

## Observations

The model converged successfully.

Validation loss decreased steadily during training.

Validation accuracy reached approximately 70%.

EarlyStopping and ModelCheckpoint ensured the best-performing model was preserved.

The baseline model is suitable for further fine-tuning and architectural improvements.

## Conclusion

A production-ready EfficientNetV2 baseline was successfully trained.

The complete training pipeline, callbacks, and history logging worked as expected.

The saved baseline model will be used in the next stage for evaluation and comparison with future models such as EfficientNet + CBAM and the Hybrid CNN–Vision Transformer architecture.
