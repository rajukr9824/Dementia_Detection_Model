```python
import tensorflow as tf
import numpy as np

from src.models.efficientnet import (
    build_efficientnet_model,
    print_model_summary,
)

from src.training.compile import compile_model
```

```python
print("TensorFlow Version:", tf.__version__)
```

```python
model = build_efficientnet_model()

print("Model built successfully.")
```

```python
print_model_summary(model)
```

```python
model = compile_model(model)

print("Model compiled successfully.")
```

```python
print("Optimizer :", model.optimizer.__class__.__name__)
print("Loss      :", model.loss.__class__.__name__)
print("Metrics   :", model.metrics_names)
```

```python
dummy_batch = tf.random.uniform(
    shape=(1, 224, 224, 3),
    dtype=tf.float32,
)

predictions = model(dummy_batch)

print("Prediction Shape:", predictions.shape)
```

```python
predictions.numpy()
```

```python
print(
    "Sum of probabilities:",
    np.sum(predictions.numpy())
)
```

```python
class_names = [
    "MildDemented",
    "ModerateDemented",
    "NonDemented",
    "VeryMildDemented",
]

predicted_index = np.argmax(predictions.numpy())

print("Predicted Class :", class_names[predicted_index])

print(
    "Confidence:",
    predictions.numpy()[0][predicted_index]
)
```
